```python
#Part 13 — Data Visualization
#Create a bar chart for Category-wise Sales.

import matplotlib.pyplot as plt

a = df.groupby("Category")["Sales"].sum()


plt.bar(a.index,a.values)
plt.xlabel("Category")
plt.ylabel("Sales")
plt.title("Category wise sales")
plt.show()
```

```python
#Create a bar chart for City-wise Profit.

a = df.groupby("City")["Profit"].sum()


plt.barh(a.index,a.values)
plt.xlabel("Profit")
plt.ylabel("City")
plt.title("City wise Profit")
plt.show()
```

```python
#Create a line chart for Year-wise Sales.
a = df.groupby("Year")["Sales"].sum()

plt.plot(a.index,a.values,color= "pink",marker="o",linestyle="dashed")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.title("Year wise sales")
plt.show()
```

```python
#Create a line chart for Month-wise Sales.
a = df.groupby("Months")["Sales"].sum()

plt.plot(a.index,a.values,marker="o",color="pink")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Monthly wise sales")
plt.show()
```

```python
#Create a bar chart for the Top 10 Cities by Sales.

a = df.groupby("City")["Sales"].sum().sort_values(ascending=False).head(10)

plt.barh(a.index,a.values,color="pink")
plt.xlabel("Sales")
plt.ylabel("City")
plt.title("Top 10 Cities by Sales")
plt.show()
```

```python
#Create a pie chart for Sales_Channel-wise Sales.
a = df.groupby("Sales_Channel")["Sales"].sum()

plt.pie(a.values,labels= a.index,autopct = '%1.1f%%',startangle = 90)
plt.title("Sales_Channel-wise Sales")

plt.show()
```
